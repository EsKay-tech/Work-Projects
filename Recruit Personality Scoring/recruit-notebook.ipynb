# This notebook aims to examine the personality scores for recruits and the respective departments that they have applied for.

## We imported the personality scores formatted as a .csv into a table

```python
import pandas as pd
import numpy as np


personality_scores = pd.read_csv('datasets/personality_scores.csv', sep = ';')
personality_scores.head()
```

### Examine the data frame for duplicates

We don't want any duplicates within our table, and each record should be unique (based on ID), therefore we need to drop duplicate rows that exist as well as any null cells. We should also become aware of a change in number of rows, these could affect our end results largely and since each ID is unique, we should be notified if a record is missing.

```python
clean_personality_scores = personality_scores.drop_duplicates('ID')
assert clean_personality_scores.shape[0] == personality_scores.ID.nunique(), 'lengths of dataframes are not the same'
```

Check for missing values.

```python
print(clean_personality_scores.isnull().sum())
```

The majority of columns have Null values for the entire column, we should remove these, they provide no value.

```python
clean_personality_scores = personality_scores.dropna(1)
clean_personality_scores
```

As seen in the previous table, the name of the headings all contain redundant strings, namely: 'Section 5 of 6 \[ \]', we need to remove them.

```python
new_columns =[]
for column in clean_personality_scores.columns:
    column = column.replace('Section 5 of 6 [', '')
    column = column.replace(']', '')
    new_columns.append(column)
clean_personality_scores.columns = new_columns
clean_personality_scores.head()
```

## Questionnaire scoring
The test consists of fifty items rated agree, neutral or disagree. To keep scoring consistent with the
original five point scale, items are scored 1=Disagree, 3=Neutral and 5=Agree.

Scores are saved as tuples. The first value of the tuple always indicates the subscale to which the item belongs:
1 = Extraversion, 2 = Agreeableness, 3 = Conscientiousness, 4 = Emotional Stability/Neuroticism, and 5 = Intellect/Imagination / openness to experiences).
The second value in the tuple indicates the individual's scored response to that question (i.e., 1=Disagree, 3=Neutral and 5=Agree).

Sum all of the individual's responses within a subscale (e.g. extraversion) to obtain a total subscale score.
In other words, the second number in the tuple should be scored for all columns with the same number in the first position (1 indicates extraversion, for example)

```python
# ignoring the Id column cos we've checked for duplicates
clean_personality_scores.set_index('ID', inplace=True)
```

### Create new columns containing the total score of each of the personality test subscales.

```python
def score_of_each_category(row):
    extraversion, agreeable, conscientiousness, emotional_stability, intellect = 0,0,0,0,0
    for cell in row:
        clean_cell = cell.strip('(')
        clean_cell = clean_cell.strip(')')
        clean_cell = clean_cell.replace(' ', '')
        
        category = int(clean_cell[0])
        
        if category == 1:
            extraversion += int(clean_cell[-1])
        elif category == 2:
            agreeable += int(clean_cell[-1])
        elif category == 3:
            conscientiousness += int(clean_cell[-1])
        elif category == 4:
            emotional_stability += int(clean_cell[-1])
        elif category == 5:
            intellect += int(clean_cell[-1])
    return extraversion, agreeable, conscientiousness, emotional_stability, intellect

score_list = [x for x in clean_personality_scores.apply(score_of_each_category, axis=1)]
score_df = pd.DataFrame(score_list, columns= ['extraversion', 'agreeable', 'conscientiousness'
                                                                            ,'emotional_stability', 'intellect'])

personality_scores = pd.concat([clean_personality_scores, score_df], axis=1)
personality_scores
```

## We want to import the department data

```python
department = pd.read_csv('datasets/departments.csv', sep=';', index_col= 'ID')
department.head()
```

We should look at the various departments the recruits have joined

```python
department.Department.unique()
```

While scanning the departments data, we realize that there are duplicates namely: 'Web Dev' and 'Web dev', this ideally should have same characters, we need to change one or the other. We remove the duplicates for correctness of our results, without correction it would mean that we have six departments when we actually have five.

```python
department['Department'] = department['Department'].str.replace('Web dev', 'Web Dev')
department.Department.unique()
```

```python
personality_scores_department = pd.concat([personality_scores, department], axis=1)

assert len((personality_scores_department)) == len((department)), 'lengths of dataframes are not the same'
assert (personality_scores_department.shape[1]) == (personality_scores.shape[1] + department.shape[1]), 'length of columns are not the same'
```

Filter the merged data frame so that you get only the applicants who scored less than 30 on emotional stability, conscientiousness AND agreeableness. Print the ID numbers and departments of these applicants to the screen, and also assign these applicants the tag “high_risk” in a new column. All other applicants get the tag “low_risk”

```python
smaller_than_thirty = personality_scores_department[(personality_scores_department['emotional_stability'] < 30) & 
                    (personality_scores_department['conscientiousness'] < 30) & (personality_scores_department['agreeable'] < 30)]
smaller_than_thirty['Department']
```

Assign the applicants 'high_risk' in a new column and the other applicants 'low_risk'

```python
personality_scores_department['Risk'] = np.where((personality_scores_department['emotional_stability'] < 30) & 
                            (personality_scores_department['conscientiousness'] < 30) & (personality_scores_department['agreeable'] < 30), 
                            'high_risk', 'low_risk')
```

### Create a new data frame with a count of the number of low and high risk applicants within each department. Let each department be a separate column.

```python
personality_scores_department[['Risk', 'Department']].pivot_table(index='Risk', columns='Department', aggfunc=len, fill_value=0)
```
